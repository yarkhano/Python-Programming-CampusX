```python
class Point:

    def __init__(self,x,y):
        self.x_cod = x
        self.y_cod = y

    def __str__(self):
        return '<{},{}>'.format(self.x_cod,self.y_cod)

    def coordinates(self,other):
        print("Distance between two coordinates is.")
        c = ((other.x_cod - self.x_cod)**2 + (other.y_cod - self.y_cod)**2)**0.5
        return c


    def dist_origin(self):
       d = (self.x_cod + self.y_cod)**0.5
       return d
```

```python
p1 = Point(2,2)
p2 = Point(3,10)
p1.coordinates(p2)
```

```python
class OnLine():

    def __init__(self,A,B,C):
        self.A = A
        self.B = B
        self.C = C

    def __str__(self):
        return '<{},{},{}>'.format(self.A,self.B,self.C)

    def find_line(line,point):
        if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
            return 'lies on the line'
        else:
            return 'does not lie on the line'
```

```python
l1 = OnLine(1,1,-2)
p1 = Point(1,1)
l1.find_line(p1)
```

## Reference: Whenever an object is created by calling its class (e.g., Person()), and we assign it to a variable, the variable stores the reference (or memory address) of the object. The variable acts as a pointer to the object in memory. If we create two variables pointing to the exact same object, any changes made through one variable will reflect in the object, affecting what the other variable sees. ##

```python
class Person:

    def __init__(self):
        self.name = ''
        self.age = 0

```

```python
p1 = Person()
p1.name = 'John'
p1.age = 10
print(p1.name)
print(p1.age)
```

**For object creation there is no need of the refrence variable but for future use of object there is need**

**Important! Any change in any object variable of the class will reflect in all objects because all variables point to the same address.**

```python
class Person:

    def __init__(self, name, gender):
        self.name = name
        self.gender = gender

    def greet(self):  # Conventionally named 'self' instead of 'person'
        self.name = 'Yar Khan'
        print(self.name)


p1 = Person('Yar', 'male')
print(p1.name)

p1.greet()  # Call it using the object instance
```

## If you modify a mutable object, it updates in place and keeps its original memory address. If you try to modify an immutable object, Python forces the creation of a new object, resulting in a completely new memory address. ##

## Instance Variable-> Variable whose value is depending or changing according to object ##

```python
class Man():

    def __init__(self,name,country):
        #instance variables
        self.name  = name
        self.country = country

#here you can see the value of variables is different for every object,instance variables
m1 = Man("Yar","Sangrai")
m2 = Man("Noor","Olandar")
print(m1.name)
print(m1.country)
print(m2.name)
print(m2.country)
```

## Encapsulation->Is wrapping or hiding or protecting the calss attributes ##

**Why the Encapsulation->So as we know that if we private the attribute in python class ,then the variable or attribute still can be used by someone to make changes outside the class using _classname__attributename ,so for every variable we create a getter method through which we access and getter to make changes outside the class.We wrapp our private variable in a method and we can add checks for validation too**

```python
#storing the objects in lists

class Person:

    def __init__(self,name,gender):
        self.name = name
        self.gender = gender


p1 = Person('Yar','male')
p2 = Person('Noor','male')
p3 = Person("Memee","Male")

L = [p1,p2,p3]

for p in L:
    print(p.name,p.gender)
```

```python
#storing the objects in dictionaries

class Person:

    def __init__(self,name,gender):
        self.name = name
        self.gender = gender


p1 = Person('Yar','male')
p2 = Person('Noor','male')
p3 = Person("Memee","Male")

d = {"p1":p1,"p2":p2,"p3":p3}
for p in d:
    print(d[p].name)
```
