```python
class Point:

    def __init__(self,x,y):
        self.x_cod = x
        self.y_cod = y

    def __str__(self):
        return '<{},{}>'.format(self.x_cod,self.y_cod)

    def coordinates(self,other):
        print("Distance between two coordinates is.")
        c = ((other.x_cod - self.x_cod)**2 + (other.y_cod - self.y_cod)**2)**0.5
        return c


    def dist_origin(self):
       d = (self.x_cod + self.y_cod)**0.5
       return d
```

```python
p1 = Point(2,2)
p2 = Point(3,10)
p1.coordinates(p2)
```

```python
class OnLine():

    def __init__(self,A,B,C):
        self.A = A
        self.B = B
        self.C = C

    def __str__(self):
        return '<{},{},{}>'.format(self.A,self.B,self.C)

    def find_line(line,point):
        if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
            return 'lies on the line'
        else:
            return 'does not lie on the line'
```

```python
l1 = OnLine(1,1,-2)
p1 = Point(1,1)
l1.find_line(p1)
```

## Refrence -> whenever object is created by calling its class e.g Person(),then when we assign this to variable the variable store the refrence of the class or address of the class.The created variable point to the object ad if we create two variables of ame object the change in any will reflect in object. ##

```python
class Person:

    def __init__(self):
        self.name = ''
        self.age = 0

```

```python
p1 = Person()
p1.name = 'John'
p1.age = 10
print(p1.name)
print(p1.age)
```

**For object creation there is no need of the refrence variable but for future use of object there s need**

**Important! Any change in any object of the class will reflect in all objects because all objects point to the same address.**

```python
class Person:

    def __init__(self, name, gender):
        self.name = name
        self.gender = gender

    def greet(self):  # Conventionally named 'self' instead of 'person'
        self.name = 'Yar Khan'
        print(self.name)


p1 = Person('Yar', 'male')
print(p1.name)

p1.greet()  # Call it using the object instance
```

## Instance Variable-> Variable whose value is depending or changing according to object ##

```python
class Man():

    def __init__(self,name,country):
        #instance variables
        self.name  = name
        self.country = country

#here you can see the value of variables is different for every object,instance variables
m1 = Man("Yar","Sangrai")
m2 = Man("Noor","Olandar")
print(m1.name)
print(m1.country)
print(m2.name)
print(m2.country)
```
