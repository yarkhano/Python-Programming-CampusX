## Polymorphism & Inheritance ##

## Aggregation-> One class own the other class. e.g Customer has Address. (has relationship) ##

**the owner class can not access the private variables of the has class directly we have to use getter method**

```python
#Aggregation

class Customer:

    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address

    def display(self):
        print(self.name)
        print(self.gender)
        print(self.address.get_city(),self.address.state,self.address.zip)

class Address:

    def __init__(self,city,state,zip):
        self.__city = city
        self.state = state
        self.zip = zip

    def __str__(self):
        return "{},{},{}".format(self.__city,self.state,self.zip)

    def get_city(self):
        return self.__city


ad1 = Address("Islamabad","E11",4400)
c1 = Customer("Yar","Male",ad1)

c1.display()
```
