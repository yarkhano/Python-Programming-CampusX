## Object Oriented programming ##

**Class is a blue print ->its a design or rules. the instance that follow these rules is called object,this represent the class**

**in classes we have variables and behavior(functions or methods)**

**constructor is a method has a super power,that it automatically called when class object is created.**

**constructor is called auto no need to call it**

**classes are of two types built in and user defined**

```python
class Atm:
    def __init__(self):
         self.pin = ''
         self.balance = 0
         self.menu()

    def menu(self):
        user_input = input(
         """
         Salam! How Can I help you?
         1.Create Pin
         2.Change Pin
         3.Withdraw Cash
         """)
        if user_input == '1':
            self.create_pin()

        elif user_input == '2':
            self.change_pin()

        elif user_input == '3':
            self.withdraw()

        else:
            pass


    def create_pin(self):
        user_pin = input("Please enter your PIN: ")
        self.pin = user_pin
        print("PIN created successfully..")

        user_balance = input("Please enter your balance: ")
        self.balance = user_balance
        print("Balance updated successfully..")

    def change_pin(self):
        old_pin = input("Please enter your Existing PIN: ")
        if  old_pin == self.pin:
            new_pin = input("Please enter your New PIN: ")
            self.pin = new_pin
            print("PIN updated successfully..")

        else:
            print("PIN not found..")


    def withdraw(self):
        user_pin = input("Please enter your PIN: ")
        if user_pin == self.pin:
            amount = input("Please enter your amount to withdraw: ")
            if amount <= self.balance:
                self.balance = self.balance - amount
                print("Withdraw successfully done..")

        else:
            print("Chal Baag")

```
