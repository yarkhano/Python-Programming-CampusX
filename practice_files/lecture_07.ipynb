```python
class Point:

    def __init__(self,x,y):
        self.x_cod = x
        self.y_cod = y

    def __str__(self):
        return '<{},{}>'.format(self.x_cod,self.y_cod)

    def coordinates(self,other):
        print("Distance between two coordinates is.")
        c = ((other.x_cod - self.x_cod)**2 + (other.y_cod - self.y_cod)**2)**0.5
        return c


    def dist_origin(self):
       d = (self.x_cod + self.y_cod)**0.5
       return d
```

```python
p1 = Point(2,2)
p2 = Point(3,10)
p1.coordinates(p2)
```

```python
class OnLine():

    def __init__(self,A,B,C):
        self.A = A
        self.B = B
        self.C = C

    def __str__(self):
        return '<{},{},{}>'.format(self.A,self.B,self.C)

    def find_line(line,point):
        if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
            return 'lies on the line'
        else:
            return 'does not lie on the line'
```

```python
l1 = OnLine(1,1,-2)
p1 = Point(1,1)
l1.find_line(p1)
```
