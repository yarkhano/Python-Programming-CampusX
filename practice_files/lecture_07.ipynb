```python
class Point:

    def __init__(self,x,y):
        self.x_cod = x
        self.y_cod = y

    def __str__(self):
        return '<{},{}>'.format(self.x_cod,self.y_cod)

    def coordinates(self,other):
        print("Distance between two coordinates is.")
        c = ((other.x_cod - self.x_cod)**2 + (other.y_cod - self.y_cod)**2)**0.5
        return c
```

```python
p1 = Point(2,2)
p2 = Point(3,10)
p1.coordinates(p2)
```
